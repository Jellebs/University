## Neural netværk
Setup: 

```python
x   = np.array([     1,    2])[:, np.newaxis]
w_1 = np.array([[    0,  0.5,  -0.5], 
                [  0.5, -0.5,  0.25], 
                [-0.25, 0.25, 0.125]])
w_2 = np.array([[-0.1],
                [ 0.4], 
                [-0.2],
                [ 0.3]])
```

```python
h1 = lambda a : 1/(1 + np.exp(-a))
h2 = lambda a : a 
```

```python
a_1 = w_1.T @ np.vstack([1, x])   # Bias added 
z_1 = h1(a_1)   
a_2 = w_2.T @ np.vstack([1, z_1]) # Bias added
y_k = h2(a_2)
```

```python
# Evaluate 
t = 1
dh1 = lambda a : (1/(1 + np.exp(-a))) * (1 - 1/(1 + np.exp(-a)))

deltak = t - y_k 
dw2 = z_1 @ deltak

deltaj = dh1(a_1) * (w_2[1:] @ deltak)      # w_2[1:] removing the bias weight. 
dw1 = np.vstack([1, x]) @ deltaj.T

# Update 
eta = 0.1
w_1 = w_1 - eta * dw1 
w_2 = w_2 - eta * np.vstack([1, dw2])       # Arbitrary dimension fix
w_2[1] = 0.1                                # Adding the bias weight back in.
```

```python
print(f"{a_1=}")
print(f"{z_1=}")
print(f"{y_k=}")
sp.pprint(f"{dw1=}")
sp.pprint(f"{dw2=}")
sp.pprint(f"{w_1=}")
sp.pprint(f"{w_2=}")
```

## Lineær regression

```python
def optimial_weights(basis, inputs, targets, M):
     # inputs (X) in R^Nx1
     # targets (t) in R^Nx1
     print(inputs)
     Phix = np.array([[basis(x, m) for m in range(M)] for x in inputs])    # Phix in R^NxM
     print(Phix)
     w = np.linalg.inv(Phix.T @ Phix) @ Phix.T @ targets                   # w in R^Mx1    
     return w

basis = lambda x, m : x**m 
x = np.array([0, 1, 3])
t = np.array([[  1],
              [  2], 
              [2.5]])

w = optimial_weights(basis, x, t, 2)
sp.pprint(f"{w=}")
```
