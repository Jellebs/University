# Helper functions

```python
def cleanLatex(tex): 
    for _ in range(2): 
        tex = tex.replace(r"\cdot", '')
        tex = tex.replace(r"\left", '')
        tex = tex.replace(r"\right", '')
        tex = tex.replace(r"j", "i")
        tex = tex.replace(r" ", "")
        tex = tex.replace(r"\!", "")
        tex = tex.replace(r"{}", "")
        tex = tex.replace(r",", ".")
        tex = tex.replace(r"\mathrm", "")
        tex = tex.replace(r"\ast", "")
        tex = tex.replace(r"\.", "")
        tex = tex.replace(r"\e", "e")
        tex = tex.replace(r"\ldots", "")
        tex = tex.replace(r"\textstyle", "")
    return tex

import subprocess

def latexDet(eq):
    pprint(eq, num_columns = 300)
    latexeq = latex(eq)
    subprocess.run("pbcopy", text=True, input=str(latexeq))

def sympyDet(tex): 
    eq = latex2sympy(tex)
    pprint(eq, num_columns = 300)
    return eq
```
