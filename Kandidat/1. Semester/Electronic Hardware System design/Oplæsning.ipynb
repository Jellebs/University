```python
f = np.array([10, 100, 1000, 10000]) * 1e6
R = (1.1592/2)*np.sqrt(f)
punkter = np.vstack([f, R])

x = np.logspace(5, 10, 1000)
y = (1.1592/2)*np.sqrt(x)
plt.xscale("log")
plt.yscale("log")
plt.plot(x, y)
plt.plot(*punkter, 'o')
```
